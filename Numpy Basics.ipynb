Q1. Is there any difference in the data type of variables list_ and array_list? If there is then write a code
to print the data types of both the variables.

Ans:- Yes, there is a difference.

list_ is a Python List, array_list is a NumPy Array (ndarray).

Q2. Write a code to print the data type of each and every element of both the variables list_ and
arra_list.

```python
import numpy as np

list_ = ['1', '2', '3', '4', '5']
array_list = np.array(object=list_)

print("Data types of elements in list_:")

for i in list_:
    print(type(i))

print("\nData types of elements in array_list:")

for i in array_list:
    print(type(i))
```

Consider the below code to answer further questions:

import numpy as np

num_list = [ [ 1 , 2 , 3 ] , [ 4 , 5 , 6 ] ]

num_array = np.array(object = num_list)

Ans:- Yes. After specifying dtype=int, all elements of array_list become integers, whereas elements of list_ remain strings.

Code:-

```python
import numpy as np

list_ = ['1', '2', '3', '4', '5']

array_list = np.array(object=list_, dtype=int)

print("list_ data types")

for i in list_:
    print(type(i))

print("\narray_list data types")

for i in array_list:
    print(type(i))
```

Q4. Write a code to find the following characteristics of variable, num_array:

(i) shape

(ii) size

```python
import numpy as np

num_list = [[1,2,3],[4,5,6]]

num_array = np.array(object=num_list)

print("Shape :", num_array.shape)
print("Size :", num_array.size)
```

Explanation:- 

Shape (2, 3) means 2 rows and 3 columns, Size 6 means total 6 elements.
